1.空间占用计算
知识点:排序、字符串、广搜、树 |来源:1.0

题目描述:


员工A 的磁盘空间经常被耗尽，他需要找到占用空间最大的目录或文件，然后决定如何清理文件释放空间.。

给定某一目录，请编写程序帮助他统计该目录内一级子目录和文件的占用空间，并返回目标目录一级子项(文件或子目录)中占用空间最大的项。

规则说明:

1.目录占用空间为其内部所有文件Size 的总和，目录本身Size 为 0.

2.目录深度不高于7 层，目录或文件名总长度不超过 128 字节。

3.当存在多个子项占用空间均为最大时，多个子项采用字符升序排列。

4.目标目录不在文件系统中时，返回空列表。

示例 1

输入:"/dir1/dir2-1",["/dir0/dir1 -1/file1-1""/dir1/dir1-1/file1-1""/dir1/dir2-1/file3-1""/dir1/dir2-1/file32","/dir1/dir2-1/dir3-1/file4-1"],[8192,81920,2048,8192,1024]

输出:["/dir1/dir2-1/file3-2"]

示例 2

输入:"/dir1",["/dir0/dir2-1/file3-1","/dir1/dir2-1/file3-1","/dir1/dir2-1/file3-2","/dir1/dir2-2/file3-3","/dir1/file2-3",[10240,4096,8192,10240,8192]

输出:["/dir1/dir2-1"]

示例 3

输入:"/dir1",["/dir1/dir1/file1","/dir1/dir1/file2","/dir1/dir2/file3"],[1024,2048,3072]

输出:["/dir1/dir1","/dir1/dir2"]




```python
from typing import List

def solution(target_dir: str, file_paths: List[str], file_sizes: List[int]) -> List[str]:
    # 统一去除目标目录末尾可能存在的斜杠
    target_dir = target_dir.rstrip('/')

    # 用于存储 目标目录下 一级子项 的累计空间大小
    # key: 子项名称, value: 累计空间
    sub_items = {}

    # 遍历文件路径及其大小
    for file_path, size in zip(file_paths, file_sizes):
        # 只处理目标目录下的文件
        if file_path.startswith(target_dir + '/'):
            # 计算相对于目标目录的路径
            relative_path = file_path[len(target_dir) + 1:]

            # 如果相对路径为空，说明是目标目录本身，根据题意忽略（目录本身大小为0）
            if not relative_path:
                continue

            # 获取该文件在目标目录下的一级子项
            if '/' in relative_path:
                direct_child = relative_path.split('/', 1)[0]
            else:
                direct_child = relative_path

            # 累加空间
            sub_items[direct_child] = sub_items.get(direct_child, 0) + size

    # 目标目录下没有一级子项，说明目标目录不在文件系统中（或为空）
    if not sub_items:
        return []

    # 找出最大的一级子项空间占用
    max_size = max(sub_items.values())

    # 找出所有占用空间为最大值的项
    result_items = [item for item, size in sub_items.items() if size == max_size]

    # 对结果按字符升序排列
    result_items.sort()

    # 返回完整的绝对路径
    return [target_dir + '/' + item for item in result_items]


solution("/dir1",["/dir0/dir2-1/file3-1","/dir1/dir2-1/file3-1","/dir1/dir2-1/file3-2","/dir1/dir2-2/file3-3","/dir1/file2-3"],[10240,4096,8192,10240,8192])
```
