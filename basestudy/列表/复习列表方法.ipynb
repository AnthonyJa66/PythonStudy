```python
lst = [1, 2, 3]

# append 是将整个列表作为一个元素塞进去
lst.append([4, 5])
print(lst)  # 输出：[1, 2, 3, [4, 5]]  （长度变成4）

lst2 = [1, 2, 3]
# extend 是把里面的元素拆开追加
lst2.extend([4, 5])
print(lst2)  # 输出：[1, 2, 3, 4, 5]  （长度变成5）
```

```python
lst = [10, 20, 30, 20]

# pop 按索引删（删掉索引为1的元素20，并返回它）
removed = lst.pop(1)
print(removed)  # 输出：20
print(lst)      # 输出：[10, 30, 20]

# remove 按值删（删掉第一个匹配的20）
lst.remove(20)
print(lst)      # 输出：[10, 30]  （注意：删的是原来的30后面的那个20，因为前面的20已经被pop删了）
```

```python
nums = [3, 1, 4, 1, 5]

# 正序排序（默认）
nums.sort()
print(nums)  # 输出：[1, 1, 3, 4, 5]

# 倒序排序
nums.sort(reverse=True)
print(nums)  # 输出：[5, 4, 3, 1, 1]

# 反转（倒序排列，不是排序）
nums.reverse()
print(nums)  # 输出：[1, 1, 3, 4, 5]（又反回来了）
```

```python
lst = ["苹果", "香蕉", "苹果", "橘子"]

# 统计个数
print(lst.count("苹果"))  # 输出：2

# 查找第一次出现的位置
print(lst.index("香蕉"))  # 输出：1

# 如果查找不存在的元素，会报错！
# lst.index("西瓜")  # 报错：ValueError: '西瓜' is not in list
```

# 真正的复制是调用copy方法，

简单赋值本质是贴标签，调用同一个列表，后续增删改查操作，列表a也会被改变。

```python
a = [1, 2, 3]
b = a          # 这不是复制，是贴标签！
b.append(4)
print(a)       # 输出：[1, 2, 3, 4]  （a 也被改了！）
```

```python
a = [1, 2, 3]
b = a.copy()   # 真正的复制
b.append(4)
print(a)       # 输出：[1, 2, 3] （a 不受影响）
```
